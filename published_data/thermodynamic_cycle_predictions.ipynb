```python
residue = "ser"

T = 310.15 #K
kB = 1.380649*10**-23 #J/K
NA = 6.022*10**23 #1/mol

Jkc = 4184 #joules per kcal
JkJ = 1000 #joules per kJ

#TODO standardize paper citations and variable names to match word document

#Tieleman and Macallum measure the free energy of bringing ethanol from water to the middle of a membrane: http://www.jgp.org/cgi/doi/10.1085/jgp.200709745
#this is analogous to the difference between the free energy of the 180 degree rotamers in those two locations
#note that this is a simulation study
dG_sidechain_group={"ser":15.8, "thr":13.9} #kJ/mol
dG_sc = dG_sidechain_group[residue]
sidechain_xfer_kT = dG_sc*JkJ/(NA*kB*T)
print(f"sidechain group transfer free energy: {sidechain_xfer_kT:.3}")

#water-membrane core transfer free energy of residues on helices:
# Hessa, T., Kim, H., Bihlmaier, K. et al. Recognition of transmembrane helices by the endoplasmic reticulum translocon. Nature 433, 377–381 (2005). https://doi.org/10.1038/nature03216
dG_residue_on_helix={"ser":0.84, "thr":0.52} #kcal/mol
dG_overall = dG_residue_on_helix[residue]
overall_xfer_kT = dG_overall*Jkc/(NA*kB*T)
print(f"overall transfer free energy: {overall_xfer_kT:.3}")

#IDK if we actually have 3 significant figures of precision here
print(f"ddG: {sidechain_xfer_kT-overall_xfer_kT:.3} kT")

```

```python
import numpy as np
#from table 2 of doi: 10.1529/biophysj.104.044024
ser_soluble_rotamer_fractions = [0.429, 0.302, 0.269]
thr_soluble_rotamer_fractions = [0.876, 0.115, 0.008]
ser_out_rotamer_counts = [26, 6, 3]
thr_out_rotamer_counts = [35, 5, 2]

def ddG(rotamer_fraction_soluble, rotamer_count_out):
    soluble_180_fraction = rotamer_fraction_soluble[2]/(rotamer_fraction_soluble[0]+rotamer_fraction_soluble[1])
    out_180_count = rotamer_count_out[2]/(rotamer_count_out[0]+rotamer_count_out[1])
    return -kB*T*np.log(out_180_count/soluble_180_fraction)

print(f"ddG for serine: {ddG(ser_soluble_rotamer_fractions, ser_out_rotamer_counts)/kB/T:.3} kT")
print(f"ddG for threonine: {ddG(thr_soluble_rotamer_fractions, thr_out_rotamer_counts)/kB/T:.3} kT")
```
